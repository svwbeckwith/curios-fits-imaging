# CuRIOS FITS Imaging - Version 2.8

This notebook demonstrates the standard workflow for analyzing a CuRIOS FITS image.

Workflow:

1. Configure analysis parameters
2. Select an observing folder and image
3. Analyze the image
4. Display the standard report
5. (Optional) Run additional diagnostics

For advanced examples, see the notebooks in the `examples/` directory.

```python
import matplotlib.pyplot as plt
import numpy as np

from fits_imaging.config import ImagingConfig
from fits_imaging.session import ImagingSession
```

```python
config = ImagingConfig()

# Optional display choices
config.colormap = "viridis"
config.contrast_method = "sigma"
config.contrast_sigma_low = 1.0
config.contrast_sigma_high = 5.0
config.stretch = "linear"

config
```

```python
session = ImagingSession()
session.status()
```

```python
results = session.analyze(config)
#results
```

```python
results.report(
    config=config,
    show_summary=True,
    show_image=True,
    show_table=True,
    show_cutouts=True,
    show_histogram=False,
    show_statistics=True,
)
```

```python
results.plot_histogram(config=config)
plt.show()
```

```python
results.source_statistics()
```

```python
results.selected_source_statistics(config=config)
```

```python
import time
import matplotlib.pyplot as plt

t = time.time()
print("1. summary")
results.print_summary()
print(f"   done in {time.time()-t:.2f} sec")

t = time.time()
print("2. statistics")
print(results.source_statistics())
print(results.selected_source_statistics(config=config))
print(f"   done in {time.time()-t:.2f} sec")

t = time.time()
print("3. main image")
fig, ax = results.plot_image(config=config, figsize=(16, 16))
plt.show()
print(f"   done in {time.time()-t:.2f} sec")

t = time.time()
print("4. peak table")
results.print_peak_table(config=config, max_rows=50)
print(f"   done in {time.time()-t:.2f} sec")

t = time.time()
print("5. cutouts")
fig, axs = results.plot_cutouts(
    config=config,
    nmax=25,
    spansize=24,
    pixel_arcsec=config.pixel_arcsec,
    exposure_sec=results.exposure_sec,
    zero_mag_counts=config.zero_mag_counts,
)
plt.show()
print(f"   done in {time.time()-t:.2f} sec")

t = time.time()
print("6. histogram")
fig, ax = results.plot_histogram(config=config)
plt.show()
print(f"   done in {time.time()-t:.2f} sec")
```
